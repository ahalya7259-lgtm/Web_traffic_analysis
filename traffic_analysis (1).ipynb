```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
print("✅ Libraries imported!")
```

```python
df = pd.read_csv('traffic.csv')
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing Values:\n", df.isnull().sum())
df.head()
```

```python
df.drop_duplicates(inplace=True)
df['date'] = pd.to_datetime(df['date'])
df['country'].fillna('Unknown', inplace=True)
df['city'].fillna('Unknown', inplace=True)
df['artist'].fillna('Unknown', inplace=True)
df['album'].fillna('Unknown', inplace=True)
df['track'].fillna('Unknown', inplace=True)
print("✅ Cleaned! Shape:", df.shape)
```

```python
print("=" * 45)
print(" 📊 KEY METRICS")
print("=" * 45)
print(f" Total Events : {len(df):,}")
print(f" Total Clicks : {(df['event']=='click').sum():,}")
print(f" Unique Artists : {df['artist'].nunique():,}")
print(f" Unique Tracks : {df['track'].nunique():,}")
print(f" Unique Countries : {df['country'].nunique():,}")
print(f" Date Range : {df['date'].min()} to {df['date'].max()}")
print("=" * 45)
```

```python
df['artist'].value_counts().head(10)\
  .plot(kind='barh', color='steelblue')
plt.title('Top 10 Artists by Clicks')
plt.xlabel('Total Clicks')
plt.tight_layout()
plt.savefig('top_artists.png')
plt.show()
```

```python
df['track'].value_counts().head(10)\
  .plot(kind='barh', color='teal')
plt.title('Top 10 Tracks by Clicks')
plt.xlabel('Total Clicks')
plt.tight_layout()
plt.savefig('top_tracks.png')
plt.show()
```

```python
df['country'].value_counts().head(10)\
  .plot(kind='bar', color='coral')
plt.title('Top 10 Countries by Clicks')
plt.xlabel('Country')
plt.ylabel('Total Clicks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('top_countries.png')
plt.show()
```

```python
df.groupby('date')['event'].count()\
  .plot(marker='o', color='steelblue')
plt.title('Daily Clicks Over Time')
plt.xlabel('Date')
plt.ylabel('Clicks')
plt.tight_layout()
plt.savefig('clicks_trend.png')
plt.show()
```

```python
df['album'].value_counts().head(10)\
  .plot(kind='barh', color='lightgreen')
plt.title('Top 10 Albums by Clicks')
plt.xlabel('Total Clicks')
plt.tight_layout()
plt.savefig('top_albums.png')
plt.show()
```

```python
df[df['city'] != 'Unknown']['city']\
  .value_counts().head(10)\
  .plot(kind='bar', color='salmon')
plt.title('Top 10 Cities by Clicks')
plt.xlabel('City')
plt.ylabel('Total Clicks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('top_cities.png')
plt.show()
```
